```python
import pandas as pd
```

```python
df = pd.read_excel("Downloads/Banking_Churn_Analysis.xlsx")

df.head()
```

```python
df['customer_id'].nunique()
```

```python
df['churn'].value_counts()
```

```python
df['churn'].mean() * 100
```

```python
pd.crosstab(df['country'], df['churn'])
```

```python
pd.crosstab(df['gender'], df['churn'])
```

```python
pd.crosstab(df['active_member'], df['churn'])
```

```python
pd.crosstab(df['products_number'], df['churn'])
```

```python
pd.crosstab(df['tenure'], df['churn'])
```

```python
df['credit_score_group'] = pd.cut(
    df['credit_score'],
    bins=[0, 600, 700, 800, 900],
    labels=['Low', 'Medium', 'Good', 'Excellent']
)

pd.crosstab(df['credit_score_group'], df['churn'])
```

```python
df['age_group'] = pd.cut(
    df['age'],
    bins=[0, 30, 40, 50, 60, 100],
    labels=['18-30', '31-40', '41-50', '51-60', '60+']
)

pd.crosstab(df['age_group'], df['churn'])
```

```python
df['salary_group'] = pd.cut(
    df['estimated_salary'],
    bins=[0, 50000, 100000, 150000, 250000],
    labels=['Low', 'Medium', 'High', 'Very High']
)

pd.crosstab(df['salary_group'], df['churn'])
```

```python
pd.crosstab(df['credit_card'], df['churn'])
```

```python
df.groupby('churn')['balance'].mean()
```

```python
df.groupby('churn')['credit_score'].mean()
```

```python
df.groupby('churn')['tenure'].mean()
```

```python
import matplotlib.pyplot as plt
```

```python
country_churn = pd.crosstab(df['country'], df['churn'])

country_churn.plot(kind='bar', figsize=(8,5))

plt.title('Country-wise Customer Churn')
plt.xlabel('Country')
plt.ylabel('Number of Customers')
plt.xticks(rotation=0)
plt.legend(['Not Churned', 'Churned'])
plt.tight_layout()
plt.show()
```
